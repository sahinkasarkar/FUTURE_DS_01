```python
import pandas as pd
```

```python
df=pd.read_excel("Superstore_Sales_Analytics.xlsx",sheet_name="Cleaned_Superstore_Data")
```

```python
df
```

```python
# Show first 5 rows
print(df.head())
```

```python
# Show column names
print(df.columns)
```

```python
# Show data info
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
# 1. Which Products Generate the Most Revenue?
top_products = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False)
print(top_products.head(10))
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
```

```python
df['Month Year'] = df['Order Date'].dt.to_period('M')
```

```python
# 2. How Do Sales Change Over Time?
monthly_sales = df.groupby('Month Year')['Sales'].sum()
print(monthly_sales)
```

```python
import matplotlib.pyplot as plt

monthly_sales.plot(kind='line')

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
# 3. Which Categories or Regions Are Most Profitable?
# Category-wise Profit
category_profit = df.groupby('Category')['Profit'].sum().sort_values(ascending=False)
print(category_profit)
```

```python
# Region-wise Profit
region_profit = df.groupby('Region')['Profit'].sum().sort_values(ascending=False)
print(region_profit)
```

```python
df.to_csv("processed_sales_data.csv", index=False)
```
